# 01 - Data Ingestion

Download and load the King County House Sales dataset from Kaggle, perform initial validation, and save raw data.

```python
# Setup
import sys
sys.path.append('../')

import pandas as pd
import numpy as np
from src.data.ingestion import (
    download_kaggle_dataset,
    load_raw_data,
    validate_data,
    save_raw_data,
    get_data_summary
)
from src.config import DATA_RAW, DATASET_NAME
```

```python
# Download dataset from Kaggle
csv_path = download_kaggle_dataset()
csv_path
```

```python
# Load and validate
df = load_raw_data(csv_path)
validation = validate_data(df)

print(f"Shape: {validation['shape']}")
print(f"Duplicate rows: {validation['duplicate_rows']}")
print(f"Duplicate IDs: {validation['duplicate_ids']}")

print("\nMissing values:")
for col, count in validation['missing_values'].items():
    if count > 0:
        print(f"  {col}: {count}")
```

```python
# Data types and basic info
print(df.dtypes)
print("\nFirst 5 rows:")
display(df.head())
```

```python
# Price statistics
if validation['price_stats']:
    print("Price Statistics:")
    for stat, val in validation['price_stats'].items():
        if isinstance(val, (int, float)):
            print(f"  {stat}: ${val:,.0f}")
        else:
            print(f"  {stat}: {val}")
```

```python
# Date range
if validation['date_range']:
    print(f"Date range: {validation['date_range']['min']} to {validation['date_range']['max']}")
```

```python
# Generate and save data summary
summary = get_data_summary(df)
summary_path = DATA_RAW / "data_summary.csv"
summary.to_csv(summary_path, index=False)
print(f"Data summary saved to: {summary_path}")
display(summary)
```

```python
# Save raw copy
save_raw_data(df)
print("Raw data saved successfully!")
```
